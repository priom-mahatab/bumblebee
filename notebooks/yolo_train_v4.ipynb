```python
import subprocess, os
out = subprocess.run(
    "find /content -path '*/drive/*' -prune -o -name dataset.yaml -print",
    shell=True, text=True, capture_output=True).stdout
DATA = out.strip().splitlines()[0]
ROOT = os.path.dirname(DATA)
subprocess.run(f"sed -i 's|^path:.*|path: {ROOT}|' '{DATA}'", shell=True)
print("DATA =", DATA)
print(open(DATA).read())
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
model.train(data=DATA, epochs=80, imgsz=960, device=0, name="imgsz960")
```
