```python
import subprocess, os
DATA = subprocess.check_output(
    "find /content/av /content/datasets -name dataset.yaml 2>/dev/null",
    shell=True, text=True).strip().splitlines()[0]
ROOT = os.path.dirname(DATA)
subprocess.run(f"sed -i 's|^path:.*|path: {ROOT}|' '{DATA}'", shell=True)
print("DATA =", DATA)
print(open(DATA).read())
```

```python
from ultralytics import YOLO
model = YOLO("yolov8n.pt")
model.train(data=DATA, epochs=80, imgsz=640, device=0)
```
